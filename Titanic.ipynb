```python
import pandas as pd
import numpy as np
```

```python
# pd calls pandas, then .read_csv('csv.csv') reads the file from the folder
train_data = pd.read_csv('train.csv')

# putting .head() shows the first five values of the dataset
train_data.head()
```

```python
# We repeat this process for test.csv
test_data = pd.read_csv('test.csv')
test_data.head()
```

```python
women = train_data.loc[train_data.Sex == 'female']["Survived"]
```

**train_data.Sex == 'female'** -> this sorts the male and the female passengers and labels males as false and females as true\
**train_data.loc [train_data.Sex == 'female']** -> this sorts the passengers so only the rows with females remain then the **["Survived"]** at the end sorts out the people that survived\
The full line of code is 'women' is the total amount of women who survived in the titanic

```python
print("Total amount of women who died on the titanic:", sum(women))
```

**sum(women)** -> since the the people who survived are labeled as 1 and the people who didn't are labeled as 0, the line sums up the total amount of (1s) which counts the total amount of women who survived

```python
print("Total amount of women on the titanic:", len(women))
```

**len(women)** -> this counts the total length (or amount of women) on the titanic regardless on whether they survived or not

```python
rate_women = sum(women)/len(women)
print(sum(women),"/",len(women),"=", rate_women)
print("% of women who survived:", 100 * rate_women)
```

**rate_women =** -> this calculates the amount of women who survived on the titanic (sum(women) over the total amount of women on the ship, showing the survival rate of women on the titanic

```python
# repeat the same process for the men on the titanic
men = train_data.loc[train_data.Sex == 'male']["Survived"]
rate_men = sum(men)/len(men)

print("% of men who survived:", 100 * rate_men)
```

Comparing the precentage of women that survived (74.20) and the percentage of men (18.89), we see that almost 75% of women survived and only 19% of men survived 

```python
from sklearn.ensemble import RandomForestClassifier
```

Imports the Random Forest classification algorithm form the scikit-learn libary

```python
y = train_data["Survived"]
```

sets y as the model that will predict whether a passenger survived or not.

```python
features = ["Pclass", "Sex", "SibSp", "Parch"]
```

the columns that will be used to predict the survival (the varibles)

```python
X = pd.get_dummies(train_data[features])
X_test = pd.get_dummies(test_data[features])
```

**pd.get_dummies** -> transforms training data and testing data into categorical data like sex into numerical dummy varibles so the algorithm can proccess it

```python
model = RandomForestClassifier(n_estimators=100, max_depth=5, random_state=1)
```

**RandomForestClassifier()** -> initializes the random forest model\
**n_estimators = 100** -> the amount of decision trees\
**max_depth = 5** -> how deep you would want the tree to go to prevent overfitting\
**random_state = 1** -> ensure consistant, reproducable results

```python
model.fit(X, y)
```

trains the model by showing the training feature (X) along with the survival outcomes (y)

```python
predictions = model.predict(X_test)
```

**model.predict()** -> runs the model using infrences only\
The whole line of code means that we are using the testing dataset (X_test) to run and predict the outcome (predictions).

```python
output = pd.DataFrame({'PassengerId': test_data.PassengerId, 'Survived': predictions})
```

**pd.DataFrame()** -> using pandas, we create a data structure (a table) to store the output\
**test_data.PassengerId** PassengerId is found by going into the test data dataset and selecting the PassengerId feature\
**predictions** The survived is calcuated in the predictions model that was made above

```python
output.to_csv('submission.csv', index=False)
print("Your submission was successfully saved!")
```
